```python
import pandas as pd
import numpy as np
```

```python
import os

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))
```

```python
df = pd.read_excel("/kaggle/input/datasets/satyasaibodireddy/dataset-for-data-analytics/Dataset for Data Analytics.xlsx")
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate OrderIDs:", df["OrderID"].duplicated().sum())
```

```python
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
```

```python
print(df["CouponCode"].isnull().sum())
```

```python
df = df.drop_duplicates()
df = df.drop_duplicates(subset="OrderID")
```

```python
print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Duplicate OrderIDs after cleaning:", df["OrderID"].duplicated().sum())
```

```python
text_cols = ["Product", "PaymentMethod", "OrderStatus", "ReferralSource", "CouponCode"]

for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
for col in ["Product", "PaymentMethod", "OrderStatus", "ReferralSource"]:
    df[col] = df[col].str.title()
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
print("Invalid dates:", df["Date"].isnull().sum())
```

```python
df["Date"] = df["Date"].dt.strftime("%Y-%m-%d")
```

```python
numeric_cols = ["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
print(df[numeric_cols].isnull().sum())
```

```python
df["CalculatedPrice"] = df["Quantity"] * df["UnitPrice"]
df["PriceMismatch"] = df["TotalPrice"] != df["CalculatedPrice"]

print("Price mismatches:", df["PriceMismatch"].sum())
```

```python
print("Quantity <= 0 :", (df["Quantity"] <= 0).sum())
print("UnitPrice <= 0 :", (df["UnitPrice"] <= 0).sum())
print("TotalPrice <= 0 :", (df["TotalPrice"] <= 0).sum())
```

```python
print("FINAL DATA QUALITY REPORT")
print("-" * 40)
print("Shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate OrderIDs:", df["OrderID"].duplicated().sum())
print("Invalid dates:", df["Date"].isnull().sum())
print("Price mismatches:", df["PriceMismatch"].sum())
```

```python
output_path = "/kaggle/working/Cleaned_Dataset_Project1.xlsx"
df.to_excel(output_path, index=False)
print("Saved to:", output_path)
```
